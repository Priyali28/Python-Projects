```python

# 1. Install the library (run this in Google Colab)
!pip install -q tiktoken

# 2. Import
import tiktoken

# 3. Enter your text
text = "Artificial Intelligence is changing the world!"

# 4. Load the tokenizer
encoding = tiktoken.get_encoding("o200k_base")

# 5. Tokenize the text
tokens = encoding.encode(text)

# 6. Display results
print("Original Text:", text)
print("\nToken IDs:", tokens)
print("\nTotal Tokens:", len(tokens))

# 7. Decode individual tokens
print("\nIndividual Tokens:")

for token_id in tokens:
    print(f"ID: {token_id} | Text: {encoding.decode([token_id])}")
```

Change the text and try these examples:

```python
examples = [
    "Hello!",
    "Hello, how are you?",
    "ChatGPT uses tokens to process text.",
    "Artificial Intelligence is fascinating!",
    "I love machine learning and Python!",
    "Supercalifragilisticexpialidocious"
]

for sentence in examples:
    token_count = len(encoding.encode(sentence))
    print(f"{sentence} → {token_count} tokens")
```

A token isn't necessarily a whole word.



A word can be split into multiple tokens.



Spaces and punctuation can also affect tokenization.



Two sentences with the same number of words can have different token counts.
